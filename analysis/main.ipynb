```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns   
```

```python
car_Data = pd.read_csv('../database/divar_206_type5_raw.csv')
car_Data.head(10)
```

```python
df = pd.DataFrame(car_Data)
df.head()
```

```python
df.info()
```

```python
import re

persian_digits = {
    ord('۰'): '0', ord('۱'): '1', ord('۲'): '2', ord('۳'): '3', ord('۴'): '4',
    ord('۵'): '5', ord('۶'): '6', ord('۷'): '7', ord('۸'): '8', ord('۹'): '9',
    ord('٠'): '0', ord('١'): '1', ord('٢'): '2', ord('٣'): '3', ord('٤'): '4',
    ord('٥'): '5', ord('٦'): '6', ord('٧'): '7', ord('٨'): '8', ord('٩'): '9'
}

df['mileage'] = (
    df['mileage']
        .astype(str)
        .str.translate(persian_digits)
        .str.replace(r'[^0-9]', '', regex=True)
        .replace('', np.nan)
        .astype('Int64')
)

df['price'] = (
    df['price'].astype(str).str.translate(persian_digits).str.replace(r'[^0-9]', '', regex=True).replace('', np.nan).astype('Int64')
)
df
```

```python
a = df['price'] > 300000000
a.sum()
```

```python
plt.scatter(data=df, x = 'mileage', y = 'price')
plt.ylim(0, 2500000000)
```

```python
sns.scatterplot(data = car_Data, x = 'mileage', y = 'price1')
```

```python
df['production_year'] = df['production_year'].astype(str).str.translate(persian_digits)
df['production_year'].replace(r'[^0-9]', '', regex=True).replace('', np.nan).astype('Int64')
df['production_year'] = df['production_year'].str[:4].astype('Int64')
df['production_year']
```

```python
df['chassis'].value_counts()
```

```python
df['engine_score'] = df['engine'].map({
    'سالم': True,
    'نیاز به تعمیر': False,
    'تعویض شده': False
})

df['chassis_score'] = df['chassis'].map({
    'سالم و پلمپ': True,
    'ضربه خورده': False
})

```

```python
body_score = {
    'سالم و بی‌خط و خش': 0,
    'خط و خش جزیی': 1,
    'صافکاری بی‌رنگ': 1,
    'رنگ‌شدگی در ۱ ناحیه': 2,
    'رنگ‌شدگی در ۲ ناحیه': 3,
    'رنگ‌شدگی در ۳ ناحیه': 4,
    'رنگ‌شدگی در ۴ ناحیه': 5,
    'رنگ‌شدگی در ۵ ناحیه': 6,
    'رنگ‌شدگی در ۶ ناحیه': 7,
    'دوررنگ': 8,
    'تمام‌رنگ': 9,
    'تصادفی': 10
}

df['body_score'] = df['body'].map(body_score)
```

```python
df1 = df[['mileage', 'price', 'production_year', 'engine_score', 'chassis_score', 'body_score']].copy()
```

```python
df['color_is_white'] = df['color'].map({
    'سفید': True
    })
df['color_is_white'].fillna(False, inplace=True)
df['color_is_white'].value_counts()
```

```python
df1 = df[['mileage', 'price', 'production_year', 'engine_score', 'chassis_score', 'body_score', 'color_is_white']].copy()
```
