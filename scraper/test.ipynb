```python

url = "https://api.divar.ir/v8/postlist/w/search"


headers = {
    "Content-Type": "application/json"
}


payload = {
    "city_ids": ["1"],
    "source_view": "FILTER",
    "pagination_data": {
        "@type": "type.googleapis.com/post_list.PaginationData",
        "page": 1,
        "layer_page": 1
    },
    "disable_recommendation": False,
    "map_state": {
        "camera_info": {
            "bbox": {}
        }
    },
    "search_data": {
        "form_data": {
            "data": {
                "brand_model": {
                    "repeated_string": {
                        "value": [
                            "Peugeot 206 5"
                        ]
                    }
                },
                "category": {
                    "str": {
                        "value": "light"
                    }
                }
            }
        }
    }
}


response = requests.post(
    url,
    headers=headers,
    json=payload
)


print(response.status_code)


data = response.json()


print(data.keys())


with open(
    "raw_response.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        data,
        f,
        ensure_ascii=False,
        indent=4
    )

widgets = data["list_widgets"]

print("Count:", len(widgets))
```

```python
import requests
import pandas as pd
import time


URL = "https://api.divar.ir/v8/postlist/w/search"


headers = {
    "Content-Type": "application/json"
}


def get_page(payload):

    response = requests.post(
        URL,
        headers=headers,
        json=payload
    )

    if response.status_code == 200:
        return response.json()

    else:
        print("Error:", response.status_code)
        return None
```

```python
def extract_posts(data):

    posts = []

    for item in data["list_widgets"]:

        # فقط آگهی‌ها
        if "data" not in item:
            continue

        post = item["data"]

        posts.append({

            "title": post.get("title"),

            "top_description": 
                post.get("top_description_text"),

            "middle_description":
                post.get("middle_description_text"),

            "bottom_description":
                post.get("bottom_description_text"),

            "token":
                post.get("token")
        })

    return posts
```

```python
def get_all_posts(first_payload, max_pages=10):

    all_posts = []

    payload = first_payload.copy()


    for page in range(max_pages):

        print("Getting page:", page+1)


        data = get_page(payload)


        if data is None:
            break


        posts = extract_posts(data)


        all_posts.extend(posts)


        print(
            "Total posts:",
            len(all_posts)
        )


        if not data["pagination"]["has_next_page"]:
            break


        # آماده کردن صفحه بعد

        payload["pagination_data"] = (
            data["pagination"]["data"]
        )


        time.sleep(1)


    return all_posts
```

```python
posts = get_all_posts(
    payload,
    max_pages=20
)
```

```python
df = pd.DataFrame(posts)

print(df.shape)

df
```
